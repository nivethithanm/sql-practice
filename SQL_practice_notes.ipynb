# DBMS

Data - Raw unprocessed facts or statistics

Information - Data that is processed to become something meaningful

Database - collection of related data in a way that is easy to access, update and manage

DBMS - MySQL, PostgreSQL, Oracle, IBM DB2 etc


## Characteristics of Database Management System
1. Data stored into Tables
2. Reduced Redundancy - DBMS follows Normalisation which divides the data in such a way that repetition is minimum.
3. Data Consistency - live data needs to updated very frequently
4. Query Language
5. Security - users and roles to protect from un-authorized access
6. Support Multiple user and Concurrent Access
7. Supports Transactions

Advantages of DBMS - retrieval, reduce redundancy, segragation of application, Integration



## Components of DBMS
- Hardware
- Software
- Data
- Procedures
- Database Access Language

```python
# https://datalemur.com/questions/sql-swapped-food-delivery
SELECT
  CASE
    WHEN (order_id = (SELECT MAX(order_id) FROM orders) AND order_id%2 = 1) THEN order_id
    WHEN (order_id%2 = 0) THEN order_id-1
    WHEN (order_id%2 = 1) THEN order_id+1
  END AS corrected_order_id,
  item
FROM orders
ORDER BY 1;
```

```python
# https://datalemur.com/questions/card-launch-success
SELECT DISTINCT
  card_name,
  FIRST_VALUE(issued_amount) OVER (
    PARTITION BY card_name
    ORDER BY make_date(issue_year, issue_month, '01')
  ) as issued_amount
FROM monthly_cards_issued
ORDER BY 2 DESC;
```

```python
# https://datalemur.com/questions/top-fans-rank
WITH artist_ranks_cte AS
(
  SELECT
    artist_name,
    DENSE_RANK() OVER(ORDER BY COUNT(songs.song_id) DESC) as artist_rank
  FROM artists
  INNER JOIN songs
    ON artists.artist_id = songs.artist_id
  INNER JOIN global_song_rank
    ON songs.song_id = global_song_rank.song_id
  WHERE global_song_rank.rank <= 10
  GROUP BY artists.artist_name
)

SELECT
  artist_name,
  artist_rank
FROM artist_ranks_cte
WHERE artist_rank <= 5;
```

```python
# https://datalemur.com/questions/histogram-users-purchases
WITH cte AS (
  SELECT
    RANK() OVER(
      PARTITION BY user_id
      ORDER BY transaction_date DESC
    ) as rank,
    user_id,
    product_id,
    transaction_date
  FROM user_transactions
)

SELECT
  transaction_date,
  user_id,
  COUNT(product_id) AS purchase_count
FROM cte
WHERE rank=1
GROUP BY transaction_date, user_id
ORDER BY transaction_date;
```

```python
# https://datalemur.com/questions/odd-even-measurements
WITH rank_cte AS (
  SELECT
    CAST(measurement_time AS date) AS measurement_day,
    RANK() OVER(
      PARTITION BY CAST(measurement_time AS date)
      ORDER BY measurement_time
    ) as rank,
    measurement_value
  FROM measurements
)

SELECT
  measurement_day,
  SUM(
    CASE
      WHEN rank%2 != 0 THEN measurement_value
    END
  ) as odd_sum,
  SUM(
    CASE
      WHEN rank%2 = 0 THEN measurement_value
    END
  ) as even_sum
FROM rank_cte
GROUP BY measurement_day
ORDER BY measurement_day;
```

```python
# https://datalemur.com/questions/sql-bloomberg-stock-performance
SELECT
  date,
  ticker,
  (close -
    LAG(close, 1) OVER(
      PARTITION BY ticker
      ORDER BY date
    )
  ) as closing_price_diff_1,
  (close -
    LAG(close, 3) OVER(
      PARTITION BY ticker
      ORDER BY date
    )
  ) as closing_price_diff_3
FROM stock_prices
LIMIT 10;
```

```python
# https://datalemur.com/questions/yoy-growth-rate
WITH cte AS (
  SELECT
    *,
    EXTRACT(YEAR FROM transaction_date) AS year,
    spend as curr_year_spend,
    LAG(spend) OVER(
      PARTITION BY product_id
      ORDER BY transaction_date
    ) as prev_year_spend
  FROM user_transactions
)

SELECT
  year,
  product_id,
  curr_year_spend,
  prev_year_spend,
  ROUND(100*(curr_year_spend-prev_year_spend)/prev_year_spend, 2) AS yoy_rate
FROM cte;
```

```python
# https://datalemur.com/questions/sql-avg-review-ratings
SELECT
  EXTRACT(MONTH FROM submit_date) as mth,
  product_id,
  ROUND(AVG(stars), 2) as avg_stars
FROM reviews
GROUP BY
  EXTRACT(MONTH FROM submit_date),
  product_id
ORDER BY mth, product_id;
```

```python
# https://datalemur.com/questions/teams-power-users
SELECT sender_id, COUNT(message_id) as message_count
FROM messages
WHERE (EXTRACT(MONTH FROM sent_date) = '8' AND EXTRACT(YEAR FROM sent_date) = '2022')
GROUP BY sender_id
ORDER BY 2 DESC
LIMIT 2;
```

```python
# https://datalemur.com/questions/duplicate-job-listings
SELECT COUNT(DISTINCT companies) as duplicate_companies
FROM (SELECT company_id
FROM job_listings
GROUP BY company_id
HAVING COUNT(job_id) > 1) as companies;
```

```python
# https://datalemur.com/questions/sql-well-paid-employees
SELECT e.employee_id, e.name
FROM employee as e
JOIN employee as m
  ON e.manager_id = m.employee_id
WHERE e.salary > m.salary;
```

```python
# https://datalemur.com/questions/final-account-balance
SELECT
  account_id,
  SUM(
    CASE
      WHEN transaction_type = 'Deposit' THEN amount
      ELSE -amount
    END
  )
FROM transactions
GROUP BY account_id;
```

```python
# https://datalemur.com/questions/click-through-rate
SELECT
  app_id,
  ROUND(100.0*
  SUM(
    CASE
      WHEN event_type='click' THEN 1 ELSE 0
    END
  )/SUM(
    CASE
      WHEN event_type='impression' THEN 1 ELSE 0
    END
  ), 2) as ctr
FROM events
WHERE EXTRACT(YEAR FROM timestamp) = '2022'
GROUP BY app_id;
```

```python
# https://datalemur.com/questions/non-profitable-drugs
SELECT
  manufacturer,
  COUNT(drug) as drug_count,
  ABS(total_sales-cogs) AS total_loss
FROM pharmacy_sales
WHERE (total_sales-cogs) < 0
GROUP BY manufacturer
ORDER BY total_loss DESC;
```

```python
# https://datalemur.com/questions/total-drugs-sales
SELECT
  manufacturer,
  CONCAT('$', ROUND(SUM(total_sales)/1000000), ' million') as sale_mil
FROM pharmacy_sales
GROUP BY manufacturer
ORDER BY SUM(total_sales) DESC, manufacturer ASC;
```

```python
# https://datalemur.com/questions/frequent-callers
SELECT COUNT(policy_holder_id) AS policy_holder_count
FROM (
  SELECT
    policy_holder_id,
    COUNT(case_id) AS call_count
  FROM callers
  GROUP BY policy_holder_id
  HAVING COUNT(DISTINCT case_id) >= 3
) AS call_records;
```

```python
# https://datalemur.com/questions/alibaba-compressed-mean
SELECT ROUND(items/orders, 1) as mean
FROM (
  SELECT
    SUM(item_count::DECIMAL*order_occurrences) AS items,
    SUM(order_occurrences) AS orders
  FROM items_per_order
) as order_items;
```

```python
# https://datalemur.com/questions/sql-ibm-db2-product-analytics
SELECT count as unique_queries, COUNT(count) as employee_count
FROM (
  SELECT
    employees.employee_id,
    SUM(
      CASE
        WHEN (EXTRACT(MONTH FROM queries.query_starttime) IN ('7', '8', '9') AND EXTRACT(YEAR FROM queries.query_starttime) = '2023') THEN 1 ELSE 0
      END
    ) as count
  FROM queries
  RIGHT JOIN employees
    ON employees.employee_id = queries.employee_id
  GROUP BY employees.employee_id
) as queries_count
GROUP BY count
ORDER BY 1;
```
